```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def getX(self):
        return self.x
    def getY(self):
        return self.y

class Line:
    def __init__(self, slope, point):
        self.slope = slope
        self.intercept = point.getY() - slope * point.getX()
    def intersects(self, other):
        if (self.slope == other.slope):
            return self.intercept == other.intercept
        return True
    def intersection(self, other):
        x = (other.intercept - self.intercept) / (self.slope - other.slope)
        y = self.slope * x + self.intercept
        return Point(x, y)
```

```python
import random as rd
import math
```

```python
total = 200000000
trials = 0
intersections = 0

for trial in range(total):
    if trial % (total / 100) == total / 100 - 1:
        if trial % (total / 10) == total / 10 - 1:
            if (trial % (total / 2) == total / 2 - 1):
                print("|")
            else:
                print("|", end = "")
        else:
            print("+", end = "")
    p1 = Point(1 - rd.random() * 2, 1 - rd.random() * 2)
    p2 = Point(1 - rd.random() * 2, 1 - rd.random() * 2)
    if ((p1.getX() ** 2 + p1.getY() ** 2) > 1 or (p2.getX() ** 2 + p2.getY() ** 2 > 1)):
        continue
    trials += 1
    l1 = Line(-p1.getX() / p1.getY(), p1)
    l2 = Line(-p2.getX() / p2.getY(), p2)
    if l1.intersects(l2):
        i = l1.intersection(l2)
        if (i.getX() ** 2 + i.getY() ** 2) < 1:
            intersections += 1
print(intersections)
print(trials)
intersections / trials
```

```python
def genMidDist():
    a1 = rd.random() * 2 * math.pi
    a2 = rd.random() * 2 * math.pi
    p1 = Point(math.cos(a1), math.sin(a1))
    p2 = Point(math.cos(a2), math.sin(a2))
    return math.sqrt(((p1.getY() + p2.getY()) / 2) ** 2 + ((p1.getX() + p2.getX()) / 2) ** 2)

def calcSize(start, delta):
    return math.pi * (2 * start * delta + delta ** 2)
```

```python
total = 50000000
intersections = 0
buckets = 400
lst = [0 for i in range(buckets)]

for trial in range(total):
    i = int(genMidDist() / (1 / buckets))
    lst[i] += 1

lst2 = [lst[i] / calcSize(i / buckets, 1 / buckets) for i in range(buckets)]
print(lst)
lst2
```
